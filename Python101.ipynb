### HW คำนวณเวลาเป็นวินาทีของเวลาต่อไปนี้โดยใช้คำสั่ง split() ช่วย (print ออกมาให้สวยงาม)

00:59:25

16:06:09

````python
time = "12:30:15"
time1 = time.split(':')
print(time1)

s = t1[0]
````

````python
time = '12:30:15'
t1 = time.split(':')        # ตัดสตริงตรง ':' ได้ list -> ['12', '30', '15']
s = t1[0]                   # '12'  (ชั่วโมง)
b = t1[1]                   # '30'  (นาที)
c = t1[2]                   # '15'  (วินาที)
total_s = int(s)*60*60 + int(b)*60 + int(c)   # แปลงเป็นเลขแล้วคำนวณวินาทีรวม
tt_s = f'{time} = {total_s} s'                # จัดรูปข้อความ output
print(tt_s)
````
